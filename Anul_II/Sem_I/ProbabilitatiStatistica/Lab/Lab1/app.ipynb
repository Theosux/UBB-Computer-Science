```python
from random import sample
from math import perm, comb
from itertools import permutations, combinations

def aranjamente(cuvant: str, k: int, numarTotal=False, aleator=False):
    if numarTotal:
        return perm(len(cuvant), k)
    elif aleator:
        return sample(cuvant, k)
    else:
        return list(permutations(cuvant, k))

def combinari(cuvant: str, k: int, numarTotal=False, aleator=False):
    if numarTotal:
        return comb(len(cuvant), k)
    elif aleator:
        return sample(cuvant, k)
    else:
        return list(combinations(cuvant, k))

print(aranjamente('word',2))
print(combinari('word',2))

```

```python
from random import random
from math import dist
from matplotlib.pyplot import axis, plot, show, title

def problema(N, cerinta):
    A, B, C, D = [0, 0], [1, 0], [1, 1], [0, 1]   # vf patrat
    E = [0.5, 0.5]                                # centru
    R = 0.5                                       # raza cerc
    nr_puncte_bune = 0

    match cerinta:
        case 1:
            for _ in range(N):
                P = [random(), random()]   # pct random
                if dist(E, P) < R:         # distanta de la punct la centru mai mica decat raza
                    nr_puncte_bune += 1    # fav
                    plot(P[0], P[1], 'ro', markersize=2)
                else:
                    plot(P[0], P[1], 'bo', markersize=2)

            print(N, nr_puncte_bune / N)   # frecventa relativa
            axis('square'); axis((0, 1, 0, 1))
            show()

        case 2:
            for _ in range(N):
                P = [random(), random()]   # pct random
                if dist(E, P) < min(dist(A, P), dist(B, P), dist(C, P), dist(D, P)):  # mai aproape de centru decat de orice varf
                    nr_puncte_bune += 1    # fav
                    plot(P[0], P[1], 'ro', markersize=2)
                else:
                    plot(P[0], P[1], 'bo', markersize=2)

            print(N, nr_puncte_bune / N)   # frecventa relativa
            axis('square'); axis((0, 1, 0, 1))
            show()

        case 3:
            for _ in range(N):
                P = [random(), random()]   # pct random
                if dist(E, P) < P[1]:      # mai aproape de centru decat de latura de jos
                    nr_puncte_bune += 1    # fav
                    plot(P[0], P[1], 'ro', markersize=2)
                else:
                    plot(P[0], P[1], 'bo', markersize=2)

            print(N, nr_puncte_bune / N)   # frecventa relativa
            axis('square'); axis((0, 1, 0, 1))
            show()

        case 4:
            M1, M2, M3, M4 = [0.5, 0], [1, 0.5], [0.5, 1], [0, 0.5]   # mijloacele laturilor
            for _ in range(N):
                P = [random(), random()]   # pct random
                # unghiul din P e obtuz daca P e in cercul cu diametrul pe latura, centru = mijlocul laturii, raza 0.5
                obtuze = 0
                for M in [M1, M2, M3, M4]:
                    if dist(M, P) < 0.5:
                        obtuze += 1
                if obtuze == 2:            # 2 triunghiuri obtuzunghice si 2 ascutitunghice
                    nr_puncte_bune += 1    # fav
                    plot(P[0], P[1], 'ro', markersize=2)
                else:
                    plot(P[0], P[1], 'bo', markersize=2)

            print(N, nr_puncte_bune / N)   # frecventa relativa
            axis('square'); axis((0, 1, 0, 1))
            show()


for N in [500, 1000, 2000]:
    problema(N, 1)
print("-------------------------")
for N in [500, 1000, 2000]:
    problema(N, 2)
print("-------------------------")
for N in [500, 1000, 2000]:
    problema(N, 3)
print("-------------------------")
for N in [500, 1000, 2000]:
    problema(N, 4)
```

```python
from itertools import product
from random import randint
from matplotlib.pyplot import plot, grid, xticks, yticks, title, show
exact = {s: 0 for s in range(3, 19)} #sume
for z in product(range(1, 7), repeat=3): #toate tripletele posibile
    exact[sum(z)] += 1 #suma

N = 100000
ss = {s: 0 for s in range(3, 19)}
for _ in range(N):
    ss[randint(1, 6) + randint(1, 6) + randint(1, 6)] += 1 #simulam zar

for s in range(3, 19):
    print(s, exact[s] / 216, ss[s] / N) #suma,probabilitate suma,frecventa relativa a sumei
```

```python
# Problema 1 a)

from scipy.stats import bernoulli, binom
from matplotlib.pyplot import bar, show, hist, grid, legend, xticks

def deplasare(nr_pasi, p):
    # fiecare pas - Bernoulli(p): 1 e dreapta (+1), 0 e stanga (-1)
    pasi = 2 * bernoulli.rvs(p, size = nr_pasi) - 1
    pozitii = [0]
    for pas in pasi:
        pozitii.append(pozitii[-1] + int(pas))
    return pozitii

print(deplasare(10, 0.5))
```

```python
# Problema 1 b)

nr_pasi, p = 20, 0.6
finale = [deplasare(nr_pasi, p)[-1] for _ in range(1000)]

hist(finale, [k - 0.5 for k in range(-nr_pasi, nr_pasi + 2)], density = True, label = 'simulare')

# K = nr pasi la dreapta - Bino(nr pasi, p), pozitia finala = 2K - nr pasi
k = range(nr_pasi + 1)
bar([2 * i - nr_pasi for i in k], binom.pmf(k, nr_pasi, p), color = 'red', width = 0.5, label = 'teoretic')

xticks(range(-nr_pasi, nr_pasi + 1, 4))
grid()
legend()
show()
```

```python
# Problema 1 c)

n = 7
# pe cerc poz = poz axa modulo n
finale = [deplasare(nr_pasi, p)[-1] % n for _ in range(1000)]

hist(finale, [k - 0.5 for k in range(n + 1)], density = True, label = 'simulare')

# acelasi K - Bino, adunam prob din acelasi nod
teoretic = [0] * n
for i in range(nr_pasi + 1):
    teoretic[(2 * i - nr_pasi) % n] += binom.pmf(i, nr_pasi, p)
bar(range(n), teoretic, color = 'red', width = 0.5, label = 'teoretic')

xticks(range(n))
grid()
legend()
show()
```

```python
# Problema 2 i)

from scipy.stats import hypergeom, geom

# X - numere ghicite pe un bilet - Hyge(6, 6, 43)
#  bilet castigator: X >= 3
p = 1 - hypergeom.cdf(2, 49, 6, 6)
print("p:", p)

# nr bilete necastigatoare pana la primul - Geo(p)
N = 1000
date = geom.rvs(p, size = N) - 1 # scadem succesul
print(date[:20])
```

```python
# Problema 2 ii)

fav = 0
for x in date:
    if x >= 10:
        fav += 1
print("P(X >= 10) estimat:", fav / N)

print("P(X >= 10) teoretic:", 1 - geom.cdf(10, p))
```

```python
# Problema 3

def citeste_cuvinte(fisier):
    with open(fisier) as f:
        return f.read().split()

spam = citeste_cuvinte('keywords_spam.txt')
ham = citeste_cuvinte('keywords_ham.txt')

cuvinte = sorted(set(spam + ham))   # W1...W14

# probabilitatile
P_spam = len(spam) / (len(spam) + len(ham))
P_ham = len(ham) / (len(spam) + len(ham))

# P(W = true | C)
P_w_spam = {w: spam.count(w) / len(spam) for w in cuvinte}
P_w_ham = {w: ham.count(w) / len(ham) for w in cuvinte}

def clasifica(email):
    cuv_email = email.split()
    p_s, p_h = P_spam, P_ham
    # independenta conditionata
    for w in cuvinte:
        if w in cuv_email:
            p_s *= P_w_spam[w]
            p_h *= P_w_ham[w]
        else:
            p_s *= 1 - P_w_spam[w]
            p_h *= 1 - P_w_ham[w]
    # daca P(E) e comun compa doar numaratorii din Bayes
    print(email, '->', p_s, p_h, 'spam' if p_s > p_h else 'ham')

clasifica("invite your friend today to click here")
clasifica("call your friend today it's urgent thank you.")
```

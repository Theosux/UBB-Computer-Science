```python
# Problema 1 i)

from random import sample

def prob_cond(nr_sim):
    countA, countB = 0, 0
    for _ in range(nr_sim):
        bile_extrase = sample(['red', 'blue', 'green'], counts=[5, 3, 2], k=3)
        if 'red' in bile_extrase:
            countA += 1
            if len(set(bile_extrase)) == 1:
                countB += 1
    return countB / countA


print(prob_cond(5000))
```

```python
# Problema 1 ii)

from math import comb


P_AsiB = comb(5, 3) / comb(10, 3)
P_A = 1 - comb(5, 3) / comb(10, 3)

print("P(B|A) teoretic:", P_AsiB / P_A)   # = 1/11
```

```python
# Problema 2

from random import randrange
from matplotlib.pyplot import bar, hist, show

data = [randrange(1, 7) for _ in range(500)]   # 500 aruncari, valori 1..6
bin_edges = [k + 0.5 for k in range(7)]

hist(data, bin_edges, density = True)

distribution = dict([(i, 1/6) for i in range(1, 7)])   # 1/6 per face
bar(distribution.keys(), distribution.values(), color = 'red')

show()
```

```python
# Problema 3 a)

from scipy.stats import binom
n = 5
p = 0.6

X = binom.rvs(n, p, size = 1000)
print(X[:20])   # primele 20
```

```python
# Problema 3 b)

from matplotlib.pyplot import bar, hist, show

hist(X, [k - 0.5 for k in range(n + 2)], density = True)

valori = range(n + 1)
bar(valori, binom.pmf(valori, n, p), color = 'red')

show()
```

```python
# Problema 3 c)

fav = 0
for x in X:
    if 1 < x <= 4:
        fav += 1
print("P(1 < X <= 4) estimat:", fav / len(X))

print("P(1 < X <= 4) teoretic:", binom.cdf(4, n, p) - binom.cdf(1, n, p))
```

```python
# Problema 4 a)

from random import randint, sample

def algoritm(v):
    k = randint(1, 100)
    w = sample(v, k)
    return max(w)

v = list(range(1, 101))
print(algoritm(v))
```

```python
# Problema 4 b)

N = 10000
fav = 0
for _ in range(N):
    if algoritm(v) == 100:
        fav += 1

print("P(output = 100) estimat:", fav / N)
```

```python
# Problema 4 c)

P = 0
for k in range(1, 101):
    P += (k / 100) * (1 / 100)

print("P(output = 100) teoretic:", P)
```
